```python
import torch
from diffusers import StableDiffusionPipeline
import gradio as gr
```

```python
model_id = "runwayml/stable-diffusion-v1-5" #Loaded a stable Diffusion model

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
) #Main AI Model - StableDiffusionPipeline

pipe = pipe.to("cuda") #GPU
```

```python
def generate_image(prompt):

    image = pipe(prompt).images[0]

    return image
```

```python
interface = gr.Interface(
    fn=generate_image,

    inputs=gr.Textbox(
        lines=2,
        placeholder="Enter your imagination here..."
    ),

    outputs=gr.Image(type="pil"),

    title="VisionCraft AI",

    description="Generate beautiful AI images using text prompts"
)

interface.launch(debug=True)
```
